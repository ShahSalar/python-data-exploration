Calculate the different joins for these two tables.

First, do them by hand


Then, confirm your results with code

```python
band_members_4rows = pd.concat([
  band_members,
  pd.DataFrame(dict(Name=["Paul"], Band=["Wings"]))
], ignore_index=True)

band_instruments_4rows = pd.concat([
    band_instruments,
    pd.DataFrame(dict(Name=["Paul"], Plays=["vocals"]))
], ignore_index=True)

inner_join_band = band_members_4rows.merge(band_instruments_4rows)
left_join_band = band_members_4rows.merge(band_instruments_4rows, on="Name", how="left")
right_join_band = band_members_4rows.merge(band_instruments_4rows, on="Name", how="right")
outer_join_band = band_members_4rows.merge(band_instruments_4rows, on="Name", how="outer")

inner_join_band, left_join_band, right_join_band, outer_join_band
```

## Part 4 - Slide 36

`horse_racing.csv` contains data related to horse racing licensing in New York.


```python
horse_racing = pd.read_csv(
    "https://raw.githubusercontent.com/ProfAvery/cpsc375-classwork/refs/heads/main/datasets/horse_racing.csv"
)
```

Transform the dataset such that:

* The License column is split into two:
  - `LicenseOrReceipt`: a categorical variable with two categories "License" and "Receipt"
  - `Number`: numeric column with the license/receipt number


```python
horse_racing["License"].head()
```

```python
horse_racing[["LicenseOrReceipt","Number"]] = horse_racing.License.str.split(' # ', expand=True)
horse_racing[["LicenseOrReceipt", "Number"]]
```

```python
horse_racing["Number"].dtype
```

```python
horse_racing["Number"] = horse_racing["Number"].astype(int)
```

```python
horse_racing["Number"].dtype
```

```python
horse_racing["LicenseOrReceipt"] = horse_racing.LicenseOrReceipt.str.capitalize()
```

```python
horse_racing["LicenseOrReceipt"].unique()
```

```python
horse_racing["License"]
```

```python
horse_racing = horse_racing.drop('License', axis=1)
```

* The `Expires` column is split into three:
  - `Year`, `Month`, `Day` (all integers)

```python
horse_racing["Expires"]
```

```python
horse_racing[["Month", "Day", "Year"]] = horse_racing.Expires.str.split("/", expand=True)
horse_racing[["Month", "Day", "Year"]]
```

```python
horse_racing[["Month", "Day", "Year"]] = horse_racing[["Month", "Day", "Year"]].astype(int)
```

```python
horse_racing["Month"].dtype
```

```python
horse_racing = horse_racing.drop("Expires", axis=1)
```

```python
horse_racing.head()
```

## Part 5 - Slide 37

Consider the dataset: `billboard`

> Hint: convert to tidy data **first**

1. Show for each track, how many weeks it spent on the chart


```python
billboard_tidy = billboard.melt(["artist", "track", "date.entered"], var_name="Week", value_name="Rank").dropna(subset=["Rank"])
```

```python
weeks = billboard_tidy.groupby(["artist", "track"])["Week"].count().reset_index(name="Weeks")
weeks
```

2. List tracks in decreasing order of number of weeks spent on the chart


```python
weeks.sort_values("Weeks", ascending=False)
```

3. Show for each track, its top rank


```python
top_rank = billboard_tidy.groupby(["artist", "track"])["Rank"].min().reset_index(name="TopRank")
top_rank
```

4. List tracks in increasing order of its top rank


```python
top_rank.sort_values("TopRank")
```

5. Show for each artist, their top rank


```python
artist_top = billboard_tidy.groupby("artist")["Rank"].min().reset_index(name="TopRank")
artist_top
```

6. List artists in increasing order of their top rank


```python
artist_top.sort_values("TopRank")
```

7. List tracks that only spent one week in the charts


```python
weeks[weeks["Weeks"] == 1]["track"]
```

8. List tracks that only spent one week in the charts along with its artist


```python
weeks[weeks["Weeks"] == 1][["track", "artist"]]
```
