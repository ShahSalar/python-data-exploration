## Question 4

```
mpg_dataset = sm.datasets.get_rdataset("mpg", "ggplot2")
mpg = mpg_dataset.data
mpg.head()
```

**Scatterplot of displ and hwy**

```
ggplot(mpg, aes(x='displ', y='hwy')) + geom_point()
```

**Same scatterplot with descriptive labels, group member title, and limits hwy (0,40), displ (0,10)**

```
(ggplot(mpg, aes(x='displ', y='hwy'))
 + geom_point()
 + labs(x='Engine Size liters',
        y='Highway fuel economy miles per gallon',
        title='Sepehr Kermanshah and Bradley Botros')
 + xlim(0, 10)
 + ylim(0, 40))
```

**Scatterplot of displ and hwy, also showing year**

```
ggplot(mpg, aes(x='displ', y='hwy', color=as_discrete('year'))) + geom_point()
```

**Scatterplot of displ and hwy, also showing year and class**

```
(ggplot(mpg, aes(x='displ', y='hwy', color='class', shape=as_discrete('year')))
 + geom_point(size=3))
```

**Load the survey data into a variable called survey using read_csv()**

```
survey = pd.read_csv("surveydataFall2026.csv")
survey.head()
```

**How many rows are there in the data?**

```
survey.shape[0]
```

There are 430 rows.

**Are there any NAs in the dataset?**

```
pd.isna(survey).sum()
```

Yes there are NAs. Columns with a count above 0 are Math 1, Statistics 2, MachineLearning 1, PlanCPSC483 12, MajorCS 2, and FamiliarPython 1.

**How many rows have at least one NA?**

```
(~survey.notna().all(axis='columns')).sum()
```

```
survey = survey.dropna()
survey.shape[0]
```

## Question 6

**Histogram of Math**

```
ggplot(survey, aes(x='Math')) + geom_histogram()
```

**Histogram with x axis breaks aligned with the bars**

```
(ggplot(survey, aes(x='Math'))
 + geom_histogram(binwidth=1, color='white')
 + scale_x_continuous(breaks=list(range(1, 11))))
```

**Boxplot of Math**

```
ggplot(survey, aes(y='Math')) + geom_boxplot()
```

**Scatterplot of Math and Statistics. What problem do you notice?**

```
ggplot(survey, aes(x='Math', y='Statistics')) + geom_point()
```

Problem: Overplotting. Both variables are whole number ratings so many students land on the exact same point. Their dots are stacked on top of each other so the plot looks like a sparse grid and hides how many students are at each spot. You cant tell a point with one student from a point with 15.

**Scatterplot with descriptive labels, group member title, limits (1,10) on both axes, and breaks aligned with the points**

```
(ggplot(survey, aes(x='Math', y='Statistics'))
 + geom_point()
 + labs(x='Math 1-10',
        y='Statistics 1-10',
        title='Sepehr Kermanshah and Bradley Botros')
 + scale_x_continuous(limits=[1, 10], breaks=list(range(1, 11)))
 + scale_y_continuous(limits=[1, 10], breaks=list(range(1, 11))))
```

**Boxplot of ComputerScience vs. Semester**

```
ggplot(survey, aes(x=as_discrete('Semester'), y='ComputerScience')) + geom_boxplot() \
    + labs(x='Semester', y='Computer Science self-rating')
```

**Visualize the two categorical variables TakenCPSC483 and PlanCPSC483**

```
(ggplot(survey, aes(x=as_discrete('TakenCPSC483'), fill=as_discrete('PlanCPSC483')))
 + geom_bar(position='dodge')
 + labs(x='Taken CPSC 483', y='Number of students', fill='Plan to take CPSC 483?'))
```
